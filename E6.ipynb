# Using Feedforward Neural Network to implement BoW and TF-IDF

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
data = {
    'text':['I love programming',
            'Python Programming is great programming',
            'Tensorflow is a powerful programming tool',
            'I enjoy machine learning',
            'AI is the future'],
    'label':['positive','positive','positive','positive','neutral']
}
```

```python
df = pd.DataFrame(data)
```

```python
label_encoder = LabelEncoder()
df['label'] = label_encoder.fit_transform(df['label'])
```

```python
label_mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
print('\nLabel Mapping:',label_mapping)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['text'],df['label'],test_size=0.2,random_state=42)
```

```python
def build_model(input_dim):
    model=Sequential([
        Dense(16, activation='relu',input_dim=input_dim),
        Dense(8, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
    return model
```

```python
bow_vectorizer =  CountVectorizer(stop_words='english')
X_train_bow = bow_vectorizer.fit_transform(X_train).toarray()
X_test_bow = bow_vectorizer.transform(X_test).toarray()
```

```python
bow_vectorizer.get_feature_names_out()
```

```python
bow_train_df = pd.DataFrame(X_train_bow, columns=bow_vectorizer.get_feature_names_out())
bow_test_df = pd.DataFrame(X_test_bow, columns=bow_vectorizer.get_feature_names_out())
```

```python
bow_model = build_model(X_train_bow.shape[1])
bow_model.fit(X_train_bow, y_train, epochs=20, batch_size=2, verbose=1)
```

```python
loss, accuracy = bow_model.evaluate(X_test_bow, y_test)
print(loss,accuracy)
```

```python
y_pred_bow = bow_model.predict(X_test_bow)
y_pred_bow = (y_pred_bow>0.5).astype(int)
y_pred_labels_bow = label_encoder.inverse_transform(y_pred_bow.flatten())
```

# tfidf

```python
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train).toarray()
X_test_tfidf = tfidf_vectorizer.transform(X_test).toarray()
```

```python
cols = tfidf_vectorizer.get_feature_names_out()
```

```python
tfidf_train_df = pd.DataFrame(X_train_tfidf, columns=cols)
tfidf_test_df = pd.DataFrame(X_test_tfidf, columns=cols)
```

```python
tfidf_model = build_model(X_train_tfidf.shape[1])
tfidf_model.fit(X_train_tfidf, y_train, epochs=20, batch_size=2, verbose=1)
```

```python
loss, accuracy = tfidf_model.evaluate(X_test_tfidf, y_test)
accuracy, loss
```

```python
y_pred_tfidf = tfidf_model.predict(X_test_tfidf)
y_pred_tfidf = (y_pred_tfidf > 0.5).astype(int)
y_pred_labels_tfidf = label_encoder.inverse_transform(y_pred_tfidf.flatten())
```

```python
print("\nActual Labels:", label_encoder.inverse_transform(y_test))
```

```python
print("Predicted Labels (BoW):", y_pred_labels_bow)
```
